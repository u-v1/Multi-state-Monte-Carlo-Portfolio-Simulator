```python
"""
The following generates 1 dummy portfolio with specified proportions of a specific asset type. Use this code if you need to test the behaviour of the model when changing 1 variable at a time,
keeping all other factors the same.
"""

import pandas as pd
import numpy as np

def generate_loan_book(total_loans=50000):
    # 1. Define splits
    n_uk_cre = int(total_loans * 0.90)
    n_other = total_loans - n_uk_cre
    other_categories = [
        ('UK', 'FI'), ('UK', 'Corporate'), 
        ('US', 'CRE'), ('US', 'FI'), ('US', 'Corporate')
    ]
    
    # 2. Create the UK CRE Data (90%)
    df_uk_cre = pd.DataFrame({
        'region': 'UK',
        'sector': 'CRE',
        'ead': 13.361e9/total_loans,
        'lgd': 0.3,
        'rsq': 0.0,
        'credit_rating': np.random.choice(['BBB'], n_uk_cre)
    })
    
    # 3. Create the Other Data (10%)
    choices = np.random.choice(len(other_categories), n_other)
    other_data = [other_categories[i] for i in choices]
    df_other = pd.DataFrame({
        'region': [x[0] for x in other_data],
        'sector': [x[1] for x in other_data],
        'ead': 13.361e9/total_loans,
        'lgd': 0.3,
        'rsq': 0.0,
        'credit_rating': np.random.choice(['BBB'], n_other)
    })
    
    # 4. Combine
    portfolio = pd.concat([df_uk_cre, df_other]).sample(frac=1).reset_index(drop=True)
    
    # 5. Add loan_id
    portfolio['loan_id'] = range(1, len(portfolio) + 1)
    
    # 6. Reorder columns as requested
    column_order = ['loan_id', 'ead', 'lgd', 'credit_rating', 'rsq', 'sector', 'region']
    portfolio = portfolio[column_order]
    
    return portfolio

# Generate and Save
portfolio = generate_loan_book(50000)
portfolio.to_csv('LoanTest14(90norsq).csv', index=False)

print("Portfolio generated with columns:")
print(portfolio.columns.tolist())
```
